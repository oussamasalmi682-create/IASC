```python
from sympy import symbols,Not,And,Implies
from sympy.logic.inference import satisfiable
```

```python
BP,RF,RB = symbols("BP RF RB")
#BP:bourage papier
#RF:retirer feuille
#RB:resolu bourage 
```

```python
fait1 = BP
```

```python
regle1 = Implies(RF,RB)
#Cette règle représente ce qui s'est passé la semaine dernière.
```

```python
test = And(fait1,regle1,Not(RB))
print(satisfiable(test))
```
