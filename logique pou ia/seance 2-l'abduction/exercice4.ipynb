```python
from sympy import symbols,Not,And,Implies
```

```python
from sympy.logic.inference import satisfiable
```

```python
RHS,CD,FP,II = symbols('RHS CD FP II ')
```

```python
regle1 = Implies(RHS,II)
```

```python
regle2 = Implies(CD,II)
```

```python
regle3 = Implies(FP,II)
```

```python
regles =[regle1,regle2,regle3]
```

```python
fait = [II]
```

```python
#Dans le chapitre précédent, nous utilisions satisfiable() pour vérifier si une conclusion était obligatoire.
#Ici, nous allons l'utiliser autrement : pour vérifier si une hypothèse est compatible avec nos connaissances.
```

```python
test1 = And(*fait,*regles,RHS)
```

```python
print(satisfiable(test1))
#Si False : l'hypothèse est incompatible avec les connaissances.
#Si un dictionnaire : l'hypothèse est possible, mais pas nécessairement vraie.
```

```python
test2 = And(*fait,*regles,CD)
print(satisfiable(test2))
```

```python
test3 = And(*fait,*regles,FP)
print(satisfiable(test3))
```

```python
V = symbols('V')
```

```python
fait.append(V)
```

```python
test4 = And(*fait,*regles,RHS)
print(satisfiable(test4))
```
