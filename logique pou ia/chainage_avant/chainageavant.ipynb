```python
from sympy import And,Not,Implies,symbols
```

```python
ORNDP,TBV,MDN,BRCH = symbols("ORNDP  TBV  MDN BRCH")
#ORNDP:lordinateur ne demare pas
#TBV:test confirme que la batterie est vide
#MDN:manque d'energie
#BRCH:brancher le chargeur 
```

```python
fait1 = ORNDP
fait2 = TBV
regle1 = Implies(TBV,ORNDP)
regle2 = Implies(MDN,BRCH)
```

```python
faits = [ORNDP,TBV]
regles = [regle1,regle2]
```

```python
for regle in regles:
    condition,conclusion = regle.args
    #regle.args : kayjbed condition et conclusion.
    if condition in faits:
        faits.append(conclusion)
print(faits)
```

```python
changement = True

while changement:
    changement = False

    for regle in regles:
        condition, conclusion = regle.args

        if condition in faits and conclusion not in faits:
            faits.append(conclusion)
            changement = True

print("Base de faits :", faits)
```
