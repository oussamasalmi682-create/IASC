```python
from sympy import symbols,Implies
```

```python
ORNDP,TBV,MDN,BRCH = symbols("ORNDP TBV MDN BRCH")
##ORNDP:lordinateur ne demare pas
#TBV:test confirme que la batterie est vide
#MDN:manque d'energie
#BRCH:brancher le chargeur 
```

```python
fait1 = ORNDP
fait2 = TBV
```

```python
regle1 = Implies(TBV,MDN)
regle2 = Implies(MDN,BRCH)
```

```python
faits = [fait1,fait2]
regles = [regle1,regle2] 
#Notre objectif est de démontrer que BRCH est vrai.
objectif = BRCH
```

```python
faits = {ORNDP, TBV}
regles = [regle1, regle2]

# Fonction qui cherche à démontrer un objectif
def chainage_arriere(objectif):

    # 1. Si l'objectif est déjà un fait connu
    if objectif in faits:
        print(objectif, "est un fait connu")
        return True

    # 2. Chercher une règle qui permet de démontrer l'objectif
    for regle in regles:
        condition, conclusion = regle.args

        if conclusion == objectif:
            print("Pour démontrer", objectif,
                  "il faut démontrer", condition)

            # 3. Vérifier le sous-objectif
            if chainage_arriere(condition):
                return True

    # 4. Aucune preuve trouvée
    return False


# Objectif final
objectif = BRCH

if chainage_arriere(objectif):
    print("Objectif démontré :", objectif)
else:
    print("Objectif non démontré :", objectif)
```
