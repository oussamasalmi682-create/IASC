```python
#exercice 3 de td 
from sympy import symbols,Not,Implies,And
from sympy.logic.inference import satisfiable
```

```python
S,H,I,inac,A,M = symbols('S H I inac A M')
```

```python
#question1
fait1 = S
fait2 = H
```

```python
regle1 = Implies(S,I)
regle2 = Implies(And(I,H),inac)
regle3 = Implies(inac,A)
```

```python
faits=[fait1,fait2]
regles=[regle1,regle2,regle3]
```

```python
#satisfiable() kayjme3 tous les faits + toutes les règles + l'hypothèse li bghina ntestiw
test1 = And(*faits,*regles,Not(I))
print(satisfiable(test1))
#False signifie que dans tous les cas qui respectent notre base de faits (BF) et notre base de règles (BR), S1 est nécessairement indisponible.
```

```python
test2 = And(*faits,*regles,Not(inac))
print(satisfiable(test2))
```

```python
test3 = And(*faits,*regles,Not(H))
print(satisfiable(test3))
```

```python
test4 = And(*faits,*regles,Not(S))
print(satisfiable(test4))
```

```python
test5 = And(*faits,*regles,Not(A))
print(satisfiable(test5))
```

```python
test6 = And(*faits,*regles,Not(M))
print(satisfiable(test6))
```
