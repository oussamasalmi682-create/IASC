```python
from sympy import symbols,Implies,And,Not
#and=et(logique) not:represente la negation(non)
#Plutôt que d'assigner une valeur fixe comme en programmation classique, l'instruction symbols("P Q") définit $P$ et $Q$ comme des symboles mathématiques abstraits (tels que $x$ et $y$) servant à structurer nos règles logiques dont la valeur ne sera testée qu'à la fin.
```

```python
from sympy.logic.inference import satisfiable
#La fonction satisfiable (de SymPy) vérifie si une expression logique est satisfaisable, c'est-à-dire si elle peut être vraie.
#Elle teste toutes les combinaisons possibles de valeurs (Vrai/Faux) pour voir s'il existe au moins un scénario où l'expression fonctionne :
#S'il existe une solution : elle retourne un dictionnaire contenant les valeurs des variables qui rendent l'expression vraie (par exemple, {P: True, Q: True}).
#S'il n'y a aucune solution (contradiction) : elle renvoie False.
```

```python
P,Q = symbols("P Q")
```

```python
regle = Implies(P,Q)
```

```python
fait = P
```

```python
# Chercher un contre-exemple à Q
test = And(fait,regle,Not(Q))
# 1. La règle est respectée
# 2. S1 est en panne (le fait P est vrai)
# 3. Mais en même temps, S1 reste disponible (c'est-à-dire Not(Q))
```

```python
print(satisfiable(test))
```
